```python
def adjacency_vector_v2(utterances_df, codes, stanza_window=4, 
                        include_self_turn=False):
    """
    개선된 adjacency vector. stanza 정의 모드 선택 가능.
    
    Parameters
    ----------
    utterances_df : DataFrame
        한 unit의 utterance들
    codes : list of str
    stanza_window : int
        이전 turn들을 몇 개까지 같은 stanza로 볼지
    include_self_turn : bool, default False
        False (rENA-compatible): 같은 turn 내 동시 등장은 카운트 안 함
        True (pyENA-original): 같은 turn 내 동시 등장도 카운트
    
    Returns
    -------
    av : np.ndarray
    pairs : list
    """
    from itertools import combinations
    
    pairs = list(combinations(codes, 2))
    pair_to_idx = {pair: i for i, pair in enumerate(pairs)}
    av = np.zeros(len(pairs))
    
    code_matrix = utterances_df[codes].values
    n_turns = len(code_matrix)
    
    for i in range(n_turns):
        current_codes = [codes[j] for j in range(len(codes)) if code_matrix[i, j] == 1]
        if not current_codes:
            continue
        
        # 이전 turn들 (현재 turn 제외)과의 co-occurrence
        window_start = max(0, i - stanza_window + 1)
        for prev_idx in range(window_start, i):  # i 미포함 (이전 turn만)
            prev_codes = [codes[j] for j in range(len(codes)) if code_matrix[prev_idx, j] == 1]
            for c_prev in prev_codes:
                for c_curr in current_codes:
                    if c_prev == c_curr:
                        continue
                    pair = (c_prev, c_curr) if (c_prev, c_curr) in pair_to_idx else (c_curr, c_prev)
                    av[pair_to_idx[pair]] += 1
        
        # 현재 turn 내 동시 등장 — 옵션
        if include_self_turn:
            for c1, c2 in combinations(current_codes, 2):
                pair = (c1, c2) if (c1, c2) in pair_to_idx else (c2, c1)
                av[pair_to_idx[pair]] += 1
    
    return av, pairs


# At01에 대해 새 함수로 재계산 (rENA-호환 모드)
at01_data = df_large[df_large['unit'] == 'At01'].sort_values('turn').reset_index(drop=True)

av_new, pairs_new = adjacency_vector_v2(
    at01_data, ref['codes'], 
    stanza_window=4, 
    include_self_turn=False  # rENA 호환
)

# 두 모드 결과 비교
av_with_self, _ = adjacency_vector_v2(
    at01_data, ref['codes'], 
    stanza_window=4, 
    include_self_turn=True  # 원래 pyENA 동작
)

# rENA 값 가져오기
rena_at01_arr = np.array(rena_at01)

# rENA 순서로 재정렬
reorder_map = [pairs_new.index(p) for p in rena_pair_order]
av_new_reordered = av_new[reorder_map]
av_with_self_reordered = av_with_self[reorder_map]

print("At01 Adjacency Vector — 세 가지 계산 비교:")
print(f"{'코드 쌍':<25}{'rENA':<10}{'pyENA(strict)':<18}{'pyENA(self)':<15}{'일치(strict)?':<10}")
print("-" * 80)
for pair, r_val, new_val, self_val in zip(rena_pair_order, rena_at01_arr, 
                                            av_new_reordered, av_with_self_reordered):
    flag = "✓" if abs(new_val - r_val) < 0.01 else "✗"
    print(f"{str(pair):<25}{r_val:<10.0f}{new_val:<18.0f}{self_val:<15.0f}{flag}")

max_diff_strict = np.max(np.abs(av_new_reordered - rena_at01_arr))
print(f"\nStrict 모드 vs rENA 최대 차이: {max_diff_strict:.6f}")

if max_diff_strict < 0.01:
    print("✓✓✓ pyENA (strict mode)가 rENA와 정확히 일치!")
else:
    print("아직 차이가 남 — 추가 디버깅 필요")
```

```python
# At01 데이터 (turn 순서대로)
at01 = df_large[df_large['unit'] == 'At01'].sort_values('turn').reset_index(drop=True)
print("At01 첫 10 turns:")
print(at01[['turn'] + ref['codes']].head(10))
print(f"\n전체 turn 수: {len(at01)}")

# 코드 매트릭스
cm = at01[ref['codes']].values
codes_list = ref['codes']

# 여러 정의 비교
def count_av(cm, codes_list, mode):
    """다양한 stanza 정의로 AV 계산"""
    from itertools import combinations
    pairs = list(combinations(codes_list, 2))
    pair_idx = {p: i for i, p in enumerate(pairs)}
    av = np.zeros(len(pairs))
    n = len(cm)
    
    for i in range(n):
        current = [codes_list[j] for j in range(len(codes_list)) if cm[i, j] == 1]
        if not current:
            continue
        
        if mode == 'strict_prev_only':
            # 이전 3 turn (현재 제외)
            window = range(max(0, i-3), i)
            for prev in window:
                prev_c = [codes_list[j] for j in range(len(codes_list)) if cm[prev, j] == 1]
                for cp in prev_c:
                    for cc in current:
                        if cp != cc:
                            p = (cp,cc) if (cp,cc) in pair_idx else (cc,cp)
                            av[pair_idx[p]] += 1
        
        elif mode == 'window_inclusive':
            # 현재 + 이전 3 turn, 각 turn은 한 번씩만 등장
            window = range(max(0, i-3), i+1)  # 현재 포함
            all_codes_in_window = set()
            for w in window:
                for j in range(len(codes_list)):
                    if cm[w, j] == 1:
                        all_codes_in_window.add(codes_list[j])
            # 현재 코드와 윈도우 내 다른 코드들의 쌍
            for cc in current:
                for other in all_codes_in_window:
                    if cc != other:
                        p = (cc, other) if (cc, other) in pair_idx else (other, cc)
                        av[pair_idx[p]] += 0.5  # 양방향 카운트되니까 절반
        
        elif mode == 'rena_style':
            # rENA 스타일 추정: 각 utterance에서, 현재 코드들과 stanza(현재 포함) 내 코드들 카운트
            # 단, 같은 turn 내 쌍은 한 번만
            window = range(max(0, i-3), i+1)
            for w in window:
                w_codes = [codes_list[j] for j in range(len(codes_list)) if cm[w, j] == 1]
                if w == i:
                    # 같은 turn: 쌍 한 번씩
                    for c1, c2 in combinations(w_codes, 2):
                        p = (c1,c2) if (c1,c2) in pair_idx else (c2,c1)
                        av[pair_idx[p]] += 1
                else:
                    # 다른 turn: 현재 코드들 × 그 turn 코드들
                    for cc in current:
                        for wc in w_codes:
                            if cc != wc:
                                p = (cc,wc) if (cc,wc) in pair_idx else (wc,cc)
                                av[pair_idx[p]] += 1
        
        elif mode == 'forward_only':
            # 미래 turn 보기 (혹시?)
            window = range(i+1, min(n, i+4))
            for fut in window:
                fut_c = [codes_list[j] for j in range(len(codes_list)) if cm[fut, j] == 1]
                for cc in current:
                    for fc in fut_c:
                        if cc != fc:
                            p = (cc,fc) if (cc,fc) in pair_idx else (fc,cc)
                            av[pair_idx[p]] += 1
        
    return av, pairs


modes = ['strict_prev_only', 'window_inclusive', 'rena_style', 'forward_only']
results = {}
for m in modes:
    av, pairs = count_av(cm, codes_list, m)
    results[m] = av

# rENA 값
rena_arr = np.array(rena_at01)
reorder_map = [pairs.index(p) for p in rena_pair_order]

print("\n비교 (모든 값은 rENA 순서로 정렬):")
print(f"{'코드 쌍':<25}{'rENA':<8}", end='')
for m in modes:
    print(f"{m[:14]:<16}", end='')
print()
print("-" * (25 + 8 + 16*len(modes)))

for i, pair in enumerate(rena_pair_order):
    print(f"{str(pair):<25}{rena_arr[i]:<8.0f}", end='')
    for m in modes:
        val = results[m][reorder_map[i]]
        match = "✓" if abs(val - rena_arr[i]) < 0.01 else " "
        print(f"{val:<5.1f}{match:<11}", end='')
    print()

# 어느 모드가 가장 잘 맞는지
print(f"\n각 모드의 rENA와의 최대 차이:")
for m in modes:
    diff = np.max(np.abs(results[m][reorder_map] - rena_arr))
    print(f"  {m}: {diff:.2f}")
```

```python
def adjacency_vector_v3(utterances_df, codes, window_size_back=4):
    """
    rENA 추정 동작: 
    - 각 utterance i에 대해, stanza = {i, i-1, ..., i-window_size_back}
    - stanza 내 각 코드 쌍이 둘 다 1이상 등장하면 +1 (binary)
    - 즉 각 utterance마다 한 번씩 stanza를 보고 그 안의 쌍을 카운트
    """
    from itertools import combinations
    pairs = list(combinations(codes, 2))
    pair_idx = {p: i for i, p in enumerate(pairs)}
    av = np.zeros(len(pairs))
    
    code_matrix = utterances_df[codes].values
    n_turns = len(code_matrix)
    
    for i in range(n_turns):
        # stanza: current + previous window_size_back
        window_start = max(0, i - window_size_back)
        stanza = code_matrix[window_start:i+1]  # current 포함
        
        # 이 utterance에서 활성 코드만 (현재 turn)
        current_codes = [codes[j] for j in range(len(codes)) if code_matrix[i, j] == 1]
        if not current_codes:
            continue
        
        # 각 현재 코드와 stanza 내 다른 코드의 binary 동시 등장
        for cc in current_codes:
            for j, other_code in enumerate(codes):
                if other_code == cc:
                    continue
                # other_code가 stanza 어디에든 1이면 카운트
                if stanza[:, j].any():
                    pair = (cc, other_code) if (cc, other_code) in pair_idx else (other_code, cc)
                    av[pair_idx[pair]] += 1
    
    # 위 로직은 (A→B)와 (B→A)를 둘 다 카운트하니까 절반
    # 하지만 정확히 그런지 검증 필요
    return av, pairs


# At01에 적용 — wb=4
av_v3, pairs_v3 = adjacency_vector_v3(at01, ref['codes'], window_size_back=4)

# 절반 처리 안 한 버전과 한 버전 둘 다 비교
av_v3_half = av_v3 / 2

reorder_map_v3 = [pairs_v3.index(p) for p in rena_pair_order]

print("Adjacency Vector v3 (utterance-level binary stanza counting):")
print(f"{'코드 쌍':<25}{'rENA(wb=4)':<13}{'v3':<10}{'v3/2':<10}{'일치?':<10}")
print("-" * 70)
for i, pair in enumerate(rena_pair_order):
    val_full = av_v3[reorder_map_v3[i]]
    val_half = av_v3_half[reorder_map_v3[i]]
    rena_v = rena_wb4_arr[i]
    match_full = "✓" if abs(val_full - rena_v) < 0.01 else " "
    match_half = "✓" if abs(val_half - rena_v) < 0.01 else " "
    print(f"{str(pair):<25}{rena_v:<13.0f}{val_full:<6.0f}{match_full:<4}{val_half:<6.1f}{match_half:<4}")

diff_full = np.max(np.abs(av_v3[reorder_map_v3] - rena_wb4_arr))
diff_half = np.max(np.abs(av_v3_half[reorder_map_v3] - rena_wb4_arr))
print(f"\nv3 (full) vs rENA: max diff = {diff_full:.2f}")
print(f"v3/2 vs rENA:      max diff = {diff_half:.2f}")
```

```python
def vector_to_ut(v):
    """rENA의 vector_to_ut (ena.cpp line 142-155)"""
    n = len(v)
    out = []
    for i in range(2, n+1):
        for j in range(0, i-1):
            out.append(v[j] * v[i-1])
    return np.array(out, dtype=float)


def get_rena_pair_order(codes):
    """rENA의 코드 쌍 순서"""
    pairs = []
    n = len(codes)
    for i in range(2, n+1):
        for j in range(0, i-1):
            pairs.append((codes[j], codes[i-1]))
    return pairs


def ref_window_df_pyena(code_matrix, window_size=4, window_forward=0, binary=True):
    """rENA의 ref_window_df 정확 구현 (ena.cpp line 205-311)"""
    df_rows, df_cols = code_matrix.shape
    n_pairs = (df_cols * (df_cols + 1)) // 2 - df_cols
    df_co_occurred = np.zeros((df_rows, n_pairs))
    
    for row in range(df_rows):
        # Stanza 범위
        earliest_row = max(0, row - (window_size - 1))
        last_row = min(df_rows - 1, row + window_forward)
        
        curr_rows = code_matrix[earliest_row:last_row+1]
        curr_rows_summed = curr_rows.sum(axis=0)
        to_ut = vector_to_ut(curr_rows_summed)
        
        # Head 차감
        curr_rows_n = curr_rows.shape[0]
        if curr_rows_n > 0 and window_size > 1 and row - 1 >= 0:
            head_rows_count = curr_rows_n - 1 - window_forward
            if head_rows_count > 0:
                curr_rows_refs = curr_rows[:head_rows_count]
                curr_row_refs_summed = curr_rows_refs.sum(axis=0)
                to_ut_refs = vector_to_ut(curr_row_refs_summed)
                to_ut = to_ut - to_ut_refs
        
        # Tail 차감
        if curr_rows_n > 0 and window_forward > 0 and last_row <= df_rows - 1:
            tail_rows_to_use = last_row - row
            if tail_rows_to_use > 0:
                curr_rows_refs = curr_rows[-tail_rows_to_use:]
                curr_row_refs_summed = curr_rows_refs.sum(axis=0)
                to_ut_refs = vector_to_ut(curr_row_refs_summed)
                to_ut = to_ut - to_ut_refs
        
        df_co_occurred[row] = to_ut
    
    # Binary clipping (마지막에 전체 매트릭스)
    if binary:
        df_co_occurred = (df_co_occurred > 0).astype(float)
    
    return df_co_occurred


# === At01 검증 ===
at01_data = df_large[df_large['unit'] == 'At01'].sort_values('turn').reset_index(drop=True)
code_matrix = at01_data[ref['codes']].values

co_occurred = ref_window_df_pyena(code_matrix, window_size=4, window_forward=0, binary=True)
print(f"co_occurred shape: {co_occurred.shape}")

# Unit AV = 모든 row의 binary co-occurrence 합
unit_av_pyena = co_occurred.sum(axis=0)

rena_pairs = get_rena_pair_order(ref['codes'])
rena_at01_arr = np.array(rena_at01)

print("\n" + "=" * 70)
print("At01 검증 — pyENA(rENA-exact) vs rENA")
print("=" * 70)
print(f"{'코드 쌍':<25}{'rENA':<10}{'pyENA':<10}{'차이':<10}{'일치?'}")
print("-" * 70)
all_match = True
for pair, r_val, py_val in zip(rena_pairs, rena_at01_arr, unit_av_pyena):
    diff = py_val - r_val
    match = "✓" if abs(diff) < 0.01 else "✗"
    if abs(diff) >= 0.01:
        all_match = False
    print(f"{str(pair):<25}{r_val:<10.0f}{py_val:<10.0f}{diff:+.0f}      {match}")

max_diff = np.max(np.abs(unit_av_pyena - rena_at01_arr))
print(f"\n최대 절대 차이: {max_diff:.6f}")
if all_match:
    print("\n✓✓✓ 완벽 일치! rENA 알고리즘을 정확히 복제함.")
else:
    print(f"\n아직 차이 있음 — 추가 디버깅 필요")
```

```python
# 모든 unit에 대해 pyENA(rENA-exact) 적용
def compute_all_avs_rena_exact(df, codes, unit_col='unit', conversation_col='conversation',
                                 window_size=4, window_forward=0, binary=True):
    """모든 unit의 AV를 rENA-exact 알고리즘으로 계산"""
    all_avs = []
    unit_ids = []
    group_ids = []
    
    # unit이 conversation을 가질 수 있으니, (unit, conversation) 조합으로 그룹화
    for unit_id in df[unit_col].unique():
        unit_data = df[df[unit_col] == unit_id].sort_values('turn').reset_index(drop=True)
        
        # conversation별로 따로 계산 후 합산
        unit_av = np.zeros((len(codes) * (len(codes)+1)) // 2 - len(codes))
        
        for conv in unit_data[conversation_col].unique():
            conv_data = unit_data[unit_data[conversation_col] == conv].sort_values('turn').reset_index(drop=True)
            cm = conv_data[codes].values
            co_occurred = ref_window_df_pyena(cm, window_size=window_size, 
                                               window_forward=window_forward, binary=binary)
            unit_av += co_occurred.sum(axis=0)
        
        all_avs.append(unit_av)
        unit_ids.append(unit_id)
        group_ids.append(unit_data['group'].iloc[0])
    
    return np.array(all_avs), unit_ids, group_ids


print("모든 90 units에 대해 pyENA(rENA-exact) 계산 중...")
pyena_avs, unit_ids_py, group_ids_py = compute_all_avs_rena_exact(
    df_large, ref['codes'], window_size=4
)
print(f"pyENA AVs shape: {pyena_avs.shape}")

# rENA 전체 결과 가져오기
r_get_all = """
code_pair_cols <- c("Data & Theory", "Data & Question", "Theory & Question",
                    "Data & Example", "Theory & Example", "Question & Example",
                    "Data & Critique", "Theory & Critique", "Question & Critique",
                    "Example & Critique")
rena_all_avs <- as.matrix(accum$connection.counts[, ..code_pair_cols])
rena_unit_ids <- as.character(accum$connection.counts$unit)
list(avs = rena_all_avs, units = rena_unit_ids)
"""

result_r = ro.r(r_get_all)
rena_avs = np.array(result_r.rx2('avs'))
rena_unit_ids = list(result_r.rx2('units'))

print(f"rENA AVs shape: {rena_avs.shape}")
print(f"rENA unit IDs (first 5): {rena_unit_ids[:5]}")

# unit 순서 정렬 (rENA의 unit 순서에 맞게 pyENA 재정렬)
py_idx_map = {uid: i for i, uid in enumerate(unit_ids_py)}
pyena_reordered = np.array([pyena_avs[py_idx_map[uid]] for uid in rena_unit_ids])

# 전체 비교
print("\n" + "=" * 70)
print("전체 검증 (90 units × 10 pairs)")
print("=" * 70)

diff_matrix = pyena_reordered - rena_avs
max_diff = np.max(np.abs(diff_matrix))
mean_diff = np.mean(np.abs(diff_matrix))
n_mismatches = np.sum(np.abs(diff_matrix) > 0.01)
total = pyena_reordered.size

print(f"전체 cells: {total}")
print(f"불일치 cells: {n_mismatches}")
print(f"일치율: {(1 - n_mismatches/total)*100:.4f}%")
print(f"최대 절대 차이: {max_diff:.6f}")
print(f"평균 절대 차이: {mean_diff:.6f}")

if n_mismatches == 0:
    print("\n" + "✓"*30)
    print("완벽 일치! 90 units × 10 pairs = 900 cells 모두 정확히 일치")
    print("✓"*30)
else:
    print(f"\n불일치 unit 처음 5개:")
    mismatch_rows = np.where(np.any(np.abs(diff_matrix) > 0.01, axis=1))[0][:5]
    for r in mismatch_rows:
        print(f"  {rena_unit_ids[r]}: max diff = {np.max(np.abs(diff_matrix[r])):.2f}")
```

```python
# 1. rENA-exact AVs (앞서 검증한 pyena_avs, 모든 90 units)
# pyena_avs는 (90, 10), unit 순서는 unit_ids_py

# unit 순서를 rENA와 맞춤
rena_set_unit_only = [u.split('::')[0] for u in rena_set_units]
py_idx = {u: i for i, u in enumerate(unit_ids_py)}
pyena_avs_aligned = np.array([pyena_avs[py_idx[u]] for u in rena_set_unit_only])

print(f"Aligned AVs shape: {pyena_avs_aligned.shape}")
print(f"AVs first row: {pyena_avs_aligned[0]}")

# 2. Sphere normalization
def sphere_normalize(av):
    norm = np.linalg.norm(av)
    return av / norm if norm > 0 else av

M_norm = np.array([sphere_normalize(row) for row in pyena_avs_aligned])

# 3. Centering
M_centered = M_norm - M_norm.mean(axis=0)

# 4. SVD
U, S, Vt = np.linalg.svd(M_centered, full_matrices=False)
pyena_svd_points = M_centered @ Vt[:2].T

print(f"\npyENA (rENA-exact AV) SVD points shape: {pyena_svd_points.shape}")
print(f"\n첫 5 units 비교:")
print(f"{'Unit':<10}{'rENA SVD1':<14}{'pyENA SVD1':<14}{'rENA SVD2':<14}{'pyENA SVD2':<14}")
print("-" * 70)
for i in range(5):
    print(f"{rena_set_unit_only[i]:<10}"
          f"{rena_points[i,0]:+.4f}       "
          f"{pyena_svd_points[i,0]:+.4f}       "
          f"{rena_points[i,1]:+.4f}       "
          f"{pyena_svd_points[i,1]:+.4f}")

# SVD sign ambiguity 처리: 부호 통일
# 각 축마다 두 좌표의 상관계수가 음수면 pyENA 부호 뒤집기
for axis in [0, 1]:
    corr = np.corrcoef(rena_points[:, axis], pyena_svd_points[:, axis])[0, 1]
    if corr < 0:
        pyena_svd_points[:, axis] *= -1
        print(f"\n  PC{axis+1}: 부호 반전 적용 (corr was {corr:+.4f})")

# 부호 통일 후 비교
print(f"\n부호 통일 후:")
print(f"{'Unit':<10}{'rENA SVD1':<14}{'pyENA SVD1':<14}{'rENA SVD2':<14}{'pyENA SVD2':<14}{'diff1':<10}{'diff2':<10}")
print("-" * 90)
for i in range(5):
    d1 = pyena_svd_points[i, 0] - rena_points[i, 0]
    d2 = pyena_svd_points[i, 1] - rena_points[i, 1]
    print(f"{rena_set_unit_only[i]:<10}"
          f"{rena_points[i,0]:+.4f}       "
          f"{pyena_svd_points[i,0]:+.4f}       "
          f"{rena_points[i,1]:+.4f}       "
          f"{pyena_svd_points[i,1]:+.4f}       "
          f"{d1:+.4f}    {d2:+.4f}")

# 전체 차이
max_diff_x = np.max(np.abs(pyena_svd_points[:, 0] - rena_points[:, 0]))
max_diff_y = np.max(np.abs(pyena_svd_points[:, 1] - rena_points[:, 1]))
print(f"\n전체 90 units 최대 차이:")
print(f"  PC1: {max_diff_x:.6f}")
print(f"  PC2: {max_diff_y:.6f}")

if max_diff_x < 1e-4 and max_diff_y < 1e-4:
    print("\n✓✓✓ 좌표까지 완벽 일치! pyENA = rENA")
else:
    print(f"\n약간의 차이 남음 — 정규화/centering 방식 미세 차이 가능성")
```

```python
# 1. rENA의 sphere-normalized + centered 입력을 우리도 동일하게 재현
#    pyena_avs가 이미 (90, 10) rENA-exact AV
#    여기에 sphere_normalize + centering을 가하면 rENA가 MR에 넣은 입력과 같음

def sphere_normalize(av):
    norm = np.linalg.norm(av)
    return av / norm if norm > 0 else av

# pyena_avs는 노트북 02에서 만든 rENA-exact AV (unit_ids_py 순서)
normalized = np.array([sphere_normalize(row) for row in pyena_avs])
centered = normalized - normalized.mean(axis=0)

print(f"centered shape: {centered.shape}")
print(f"centered mean (≈0이어야): {centered.mean(axis=0).round(6)}")
print()

# 2. 본인 노트북 01의 means_rotation 함수 (검증할 알고리즘)
def means_rotation(M_centered, group_labels):
    M = np.asarray(M_centered, dtype=float)
    labels = np.asarray(group_labels)
    groups = np.unique(labels)
    assert len(groups) == 2, f"means_rotation은 정확히 2개 그룹 필요. 받은 그룹: {list(groups)}"
    
    g1_mean = M[labels == groups[0]].mean(axis=0)
    g2_mean = M[labels == groups[1]].mean(axis=0)
    
    diff = g1_mean - g2_mean
    x_axis = diff / np.linalg.norm(diff)
    
    projection_on_x = M @ x_axis[:, np.newaxis] @ x_axis[np.newaxis, :]
    M_residual = M - projection_on_x
    _, _, Vt_res = np.linalg.svd(M_residual, full_matrices=False)
    y_axis = Vt_res[0]
    
    rotation_basis = np.vstack([x_axis, y_axis])
    points = M @ rotation_basis.T
    return points, rotation_basis

# 3. rENA가 회전에 쓴 그룹만 추출 (A_tight, B_tight)
# unit_ids_py 순서 그대로 group 라벨 만들기
group_by_unit = {}
for unit, grp in zip(rena_mr_units, rena_mr_groups):
    unit_clean = unit.split('::')[0]
    group_by_unit[unit_clean] = grp

groups_py = np.array([group_by_unit[u] for u in unit_ids_py])
print(f"unit_ids_py 첫 5: {unit_ids_py[:5]}")
print(f"groups_py 첫 5:   {list(groups_py[:5])}")
print(f"그룹 분포: {dict(zip(*np.unique(groups_py, return_counts=True)))}")
print()

# A_tight + B_tight만 골라서 회전 적용
mask_ab = np.isin(groups_py, ['A_tight', 'B_tight'])
print(f"A_tight + B_tight 단위 수: {mask_ab.sum()}")

# 우리 means_rotation은 두 그룹만으로 회전 축을 계산
points_pyena_mr_60, basis = means_rotation(centered[mask_ab], groups_py[mask_ab])

# 4. 그런데 rENA는 90 units 전체를 회전된 축에 사영함 (A_loose도 좌표가 있었음)
# 그러니 basis로 90 units 전체를 재사영
points_pyena_mr_90 = centered @ basis.T
print(f"\npyena MR points (90 units): {points_pyena_mr_90.shape}")

# 5. rENA 결과와 매칭 — rENA unit 순서 기준으로 정렬
rena_unit_clean = [u.split('::')[0] for u in rena_mr_units]
py_idx = {u: i for i, u in enumerate(unit_ids_py)}
pyena_aligned = np.array([points_pyena_mr_90[py_idx[u]] for u in rena_unit_clean])

# 6. 비교 (sign-align 후)
def align_sign(a, b):
    out = a.copy()
    for k in range(a.shape[1]):
        if np.dot(a[:, k], b[:, k]) < 0:
            out[:, k] = -a[:, k]
    return out

pyena_signed = align_sign(pyena_aligned, rena_mr_points)
diff = np.abs(pyena_signed - rena_mr_points)

print()
print("=== rENA MR vs pyENA means_rotation 비교 ===")
print(f"shape:        {pyena_signed.shape} vs {rena_mr_points.shape}")
print(f"max |Δ| MR1:  {diff[:, 0].max():.6e}")
print(f"max |Δ| SVD2: {diff[:, 1].max():.6e}")
print(f"mismatched (>1e-10): {(diff > 1e-10).sum()} / {diff.size}")
print()
print("첫 5 units 비교:")
print(f"{'unit':<8}{'rENA MR1':>12}{'pyENA':>12}{'rENA SVD2':>14}{'pyENA':>12}")
for i in range(5):
    print(f"{rena_unit_clean[i]:<8}"
          f"{rena_mr_points[i,0]:+12.6f}{pyena_signed[i,0]:+12.6f}"
          f"{rena_mr_points[i,1]:+14.6f}{pyena_signed[i,1]:+12.6f}")
```

```python
# 가설 (b): rENA의 SVD2는 원본 centered 행렬의 SVD 두 번째 주성분 그대로?
# 즉 회전 없이 원본 SVD2를 가져온 것?

# 노트북 02의 일반 SVD 결과 (이미 검증된 것)
# centered → U, S, Vt → SVD coords = centered @ Vt.T
U_full, S_full, Vt_full = np.linalg.svd(centered, full_matrices=False)
svd_coords = centered @ Vt_full.T  # (90, 10)

# rENA가 set_mr에서 만든 SVD2가 일반 SVD의 SVD2와 같은지?
# rENA 좌표 90개를 unit_ids_py 순서로 정렬
rena_aligned_by_pyunit = np.array([rena_mr_points[rena_unit_clean.index(u)] for u in unit_ids_py])

# 일반 SVD의 두 번째 컬럼과 비교 (sign-align)
svd2_py = svd_coords[:, 1]
svd2_rena = rena_aligned_by_pyunit[:, 1]
if np.dot(svd2_py, svd2_rena) < 0:
    svd2_py_signed = -svd2_py
else:
    svd2_py_signed = svd2_py

diff_b = np.abs(svd2_py_signed - svd2_rena)
print("=== 가설 (b): rENA SVD2 == 일반 centered SVD의 SVD2? ===")
print(f"max |Δ|: {diff_b.max():.6e}")
print(f"mismatched (>1e-10): {(diff_b > 1e-10).sum()} / 90")
print()
print("첫 5 units:")
print(f"{'unit':<8}{'rENA SVD2':>14}{'pure SVD2':>14}")
for i in range(5):
    print(f"{unit_ids_py[i]:<8}{svd2_rena[i]:+14.6f}{svd2_py_signed[i]:+14.6f}")

# 일반 SVD의 다른 컬럼과도 비교 (혹시 SVD3, SVD4가 매칭일 수도)
print()
print("=== 보너스: rENA SVD2가 일반 SVD 어느 컬럼과 가장 가까운지? ===")
for k in range(min(10, Vt_full.shape[0])):
    col = svd_coords[:, k]
    if np.dot(col, svd2_rena) < 0:
        col = -col
    d = np.abs(col - svd2_rena).max()
    corr = np.corrcoef(col, svd2_rena)[0, 1]
    print(f"  pure SVD{k+1}: max|Δ| = {d:.4e}, |corr| = {abs(corr):.4f}")
```

```python
# 가설 (c): 일반 SVD의 Vt[1]에서 X축 성분 제거 후 단위화한 게 rENA의 두 번째 축?

# X축 (이미 검증된 MR 방향)
x_axis = basis[0]  # (10,) — 위에서 계산한 그룹 차이 단위벡터

# 일반 SVD의 두 번째 주성분
v2_original = Vt_full[1]  # (10,)

# Gram-Schmidt: v2에서 x_axis 방향 성분 제거
v2_orthogonal = v2_original - np.dot(v2_original, x_axis) * x_axis
v2_orthogonal = v2_orthogonal / np.linalg.norm(v2_orthogonal)

# 직교 확인
print(f"x_axis · v2_orthogonal = {np.dot(x_axis, v2_orthogonal):.2e}  (≈0이어야)")
print()

# 이 축에 centered를 사영한 게 rENA의 SVD2와 일치하는가?
svd2_proj_orthogonal = centered @ v2_orthogonal

# sign-align
if np.dot(svd2_proj_orthogonal, svd2_rena) < 0:
    svd2_proj_orthogonal = -svd2_proj_orthogonal

diff_c = np.abs(svd2_proj_orthogonal - svd2_rena)
print(f"=== 가설 (c): Gram-Schmidt 직교화된 SVD2 ===")
print(f"max |Δ|: {diff_c.max():.6e}")
print(f"mismatched (>1e-10): {(diff_c > 1e-10).sum()} / 90")
print()
print("첫 5 units:")
print(f"{'unit':<8}{'rENA SVD2':>14}{'가설(c)':>14}")
for i in range(5):
    print(f"{unit_ids_py[i]:<8}{svd2_rena[i]:+14.6f}{svd2_proj_orthogonal[i]:+14.6f}")
```

```python
def orthogonal_svd(data, weights):
    """rENA의 orthogonal_svd (rENA/R/RotationSet.R) NumPy 이식.
    
    weights를 첫 ncol(weights)개 축으로 고정한 정규직교 기저를 만들고,
    나머지 부분공간에서 SVD를 돌려 회전 행렬을 반환.
    
    data    : (n, p)  centered matrix
    weights : (p, k)  회전 축들 (means rotation은 k=1)
    returns : (p, p)  full rotation matrix
    """
    Q, _ = np.linalg.qr(weights, mode='complete')  # (p, p), Q[:,:k] == weights (정규화 후)
    k = weights.shape[1]
    
    # weights에 직교한 부분공간으로 사영
    X_bar = data @ Q[:, k:]  # (n, p-k)
    
    # prcomp(X_bar, scale=F)$rotation — center=TRUE가 디폴트
    X_bar_centered = X_bar - X_bar.mean(axis=0)
    _, _, Vt = np.linalg.svd(X_bar_centered, full_matrices=False)
    V = Vt.T  # prcomp$rotation은 V (Vt 아님)
    
    # cbind(Q[:,:k], Q[:,k:] @ V)
    return np.hstack([Q[:, :k], Q[:, k:] @ V])


def means_rotation_rena(M_centered, group_labels, group_a, group_b):
    """rENA-exact means rotation (ena.rotate.by.mean 정확 재현).
    
    Returns
    -------
    points : (n, p) 회전된 좌표 전체
    rotation : (p, p) 회전 행렬
    """
    M = np.asarray(M_centered, dtype=float)
    labels = np.asarray(group_labels)
    
    # 라인 22: scale(data, scale=F, center=T) — 다시 centering
    M = M - M.mean(axis=0)
    
    # 라인 39-44: 그룹 평균 차의 단위벡터
    g1 = M[labels == group_a].mean(axis=0)
    g2 = M[labels == group_b].mean(axis=0)
    diff = g1 - g2
    col_mean_diff_sq = diff / np.sqrt(np.sum(diff ** 2))  # (p,)
    weights = col_mean_diff_sq[:, np.newaxis]  # (p, 1)
    
    # 라인 45-46: deflation
    deflated = M - (M @ weights) @ weights.T
    
    # 라인 50: orthogonal_svd
    rotation = orthogonal_svd(deflated, weights)
    
    # 좌표 = centered_data @ rotation
    points = M @ rotation
    return points, rotation


# === 호출 + 비교 ===
points_rena_exact, rot_matrix = means_rotation_rena(
    centered, groups_py, 'A_tight', 'B_tight'
)
print(f"점 shape: {points_rena_exact.shape}")
print(f"rotation shape: {rot_matrix.shape}")
print()

# rENA 결과(center.align.to.origin=FALSE 버전)와 비교
# MR1, SVD2 = points의 첫 두 컬럼
pyena_mr_signed = align_sign(points_rena_exact[:, :2], rena_raw_aligned)

diff = np.abs(pyena_mr_signed - rena_raw_aligned)
print("=== rENA-exact means rotation 비교 ===")
print(f"MR1   max |Δ|: {diff[:, 0].max():.6e}")
print(f"SVD2  max |Δ|: {diff[:, 1].max():.6e}")
print(f"mismatched (>1e-10): MR1={(diff[:,0]>1e-10).sum()}, SVD2={(diff[:,1]>1e-10).sum()}")
print()
print("첫 5 units:")
print(f"{'unit':<8}{'rENA MR1':>12}{'pyENA':>12}{'rENA SVD2':>14}{'pyENA':>12}")
for i in range(5):
    print(f"{unit_ids_py[i]:<8}"
          f"{rena_raw_aligned[i,0]:+12.6f}{pyena_mr_signed[i,0]:+12.6f}"
          f"{rena_raw_aligned[i,1]:+14.6f}{pyena_mr_signed[i,1]:+12.6f}")
```

```python
%%writefile ../src/pyena/stats/reproducibility.py
"""ENA 공간에서 unit들의 reproducibility(수렴도) 정량화.

각 그룹(또는 전체)에 대해 4가지 지표를 계산:
  1. Centroid dispersion : 각 점에서 centroid까지의 평균 거리 (낮을수록 수렴)
  2. Pairwise distance   : 모든 점 쌍의 평균/최대 거리
  3. 95% Confidence ellipse area : 다변량 정규성 가정 하의 신뢰영역 면적
  4. Convex hull area   : 실제 점들이 차지하는 다각형 면적

낮을수록 수렴, 높을수록 발산. LLM dialogue reproducibility 분석의 기본 지표.
"""

from __future__ import annotations

import numpy as np
import pandas as pd
from scipy.spatial import ConvexHull
from scipy.spatial.distance import pdist
from scipy.stats import chi2

__all__ = ["reproducibility_metrics"]


def reproducibility_metrics(points, group_labels=None):
    """4가지 reproducibility 지표를 계산.

    Parameters
    ----------
    points : np.ndarray (n_units, 2)
        ENA 2D 좌표.
    group_labels : array-like or None, default None
        그룹별로 계산. None이면 전체 데이터에 적용.

    Returns
    -------
    pd.DataFrame
        그룹(또는 'all')별 9개 컬럼:
          n, centroid_x, centroid_y,
          centroid_dispersion_mean, centroid_dispersion_std,
          pairwise_distance_mean, pairwise_distance_max,
          ellipse_95_area, convex_hull_area
    """
    def _compute_one_group(pts):
        n = len(pts)
        centroid = pts.mean(axis=0)

        # 1. Centroid dispersion
        distances_to_centroid = np.linalg.norm(pts - centroid, axis=1)
        centroid_dispersion = distances_to_centroid.mean()
        centroid_dispersion_std = distances_to_centroid.std()

        # 2. Pairwise distance
        if n >= 2:
            pairwise = pdist(pts)
            pairwise_mean = pairwise.mean()
            pairwise_max = pairwise.max()
        else:
            pairwise_mean = np.nan
            pairwise_max = np.nan

        # 3. 95% Confidence ellipse area
        # 면적 = π × χ²(0.95, df=2) × √(λ₁ × λ₂)
        if n >= 3:
            cov = np.cov(pts.T)
            eigenvals = np.linalg.eigvalsh(cov)
            chi2_val = chi2.ppf(0.95, df=2)
            ellipse_area = np.pi * chi2_val * np.sqrt(np.prod(eigenvals))
        else:
            ellipse_area = np.nan

        # 4. Convex hull area (2D에서 ConvexHull.volume = area)
        if n >= 3:
            try:
                hull = ConvexHull(pts)
                hull_area = hull.volume
            except Exception:
                hull_area = np.nan
        else:
            hull_area = np.nan

        return {
            'n': n,
            'centroid_x': centroid[0], 'centroid_y': centroid[1],
            'centroid_dispersion_mean': centroid_dispersion,
            'centroid_dispersion_std': centroid_dispersion_std,
            'pairwise_distance_mean': pairwise_mean,
            'pairwise_distance_max': pairwise_max,
            'ellipse_95_area': ellipse_area,
            'convex_hull_area': hull_area,
        }

    if group_labels is None:
        return pd.DataFrame([_compute_one_group(points)], index=['all'])

    group_labels = np.array(group_labels)
    results = {}
    for g in np.unique(group_labels):
        pts_g = points[group_labels == g]
        results[g] = _compute_one_group(pts_g)

    return pd.DataFrame(results).T
```

```python
normalize_test = '''"""Tests for pyena.core.normalize."""

import numpy as np

from pyena.core.normalize import sphere_normalize


# ===== Reference regression =====

def test_sphere_normalize_matches_reference(pyena_reference):
    """Same AV matrix must yield the same sphere-normalized matrix as notebook 01."""
    normalized = sphere_normalize(pyena_reference["av_matrix"])
    diff = np.abs(normalized - pyena_reference["normalized_matrix"])
    assert diff.max() == 0.0


# ===== Mathematical properties =====

class TestSphereNormalize:
    def test_2d_each_row_has_unit_norm(self):
        rng = np.random.default_rng(0)
        av = rng.random((50, 10))
        out = sphere_normalize(av)
        norms = np.linalg.norm(out, axis=1)
        np.testing.assert_allclose(norms, np.ones(50), atol=1e-14)

    def test_1d_input_returns_unit_vector(self):
        v = np.array([3.0, 4.0])  # ||v|| = 5
        out = sphere_normalize(v)
        np.testing.assert_allclose(out, [0.6, 0.8])
        assert np.isclose(np.linalg.norm(out), 1.0)

    def test_zero_row_returned_unchanged(self):
        """Zero-norm rows must be passed through (rENA convention)."""
        av = np.array([[3.0, 4.0], [0.0, 0.0], [1.0, 0.0]])
        out = sphere_normalize(av)
        np.testing.assert_array_equal(out[1], [0.0, 0.0])  # unchanged
        np.testing.assert_allclose(out[0], [0.6, 0.8])
        np.testing.assert_allclose(out[2], [1.0, 0.0])

    def test_zero_vector_1d_returned_unchanged(self):
        out = sphere_normalize(np.zeros(5))
        np.testing.assert_array_equal(out, np.zeros(5))

    def test_idempotent(self):
        """Normalizing an already-normalized vector is a no-op."""
        rng = np.random.default_rng(1)
        av = rng.random((20, 8))
        once = sphere_normalize(av)
        twice = sphere_normalize(once)
        np.testing.assert_allclose(once, twice, atol=1e-14)

    def test_preserves_direction(self):
        """sphere_normalize(α·v) must equal sphere_normalize(v) for α > 0."""
        rng = np.random.default_rng(2)
        v = rng.random(10)
        out1 = sphere_normalize(v)
        out2 = sphere_normalize(5.0 * v)
        np.testing.assert_allclose(out1, out2, atol=1e-14)

    def test_shape_preserved(self):
        for shape in [(10,), (5, 3), (90, 10), (1, 5)]:
            av = np.random.default_rng(3).random(shape)
            out = sphere_normalize(av)
            assert out.shape == shape

    def test_dtype_is_float(self):
        av = np.array([[1, 2], [3, 4]], dtype=int)
        out = sphere_normalize(av)
        assert out.dtype == np.float64
'''

path = PROJECT_ROOT / 'tests' / 'test_normalize.py'
path.write_text(normalize_test)

import subprocess
result = subprocess.run(
    ['pytest', str(path), '-v'],
    capture_output=True, text=True, cwd=str(PROJECT_ROOT),
)
print(result.stdout[-2500:])
if result.stderr:
    print("=== STDERR ===")
    print(result.stderr[-500:])
print(f"return code: {result.returncode}")
```

```python
projection_test = '''"""Tests for pyena.core.projection."""

import numpy as np
import pytest

from pyena.core.projection import (
    MeansRotationResult,
    ProjectionResult,
    means_rotation,
    orthogonal_svd,
    svd_project,
)


# ===== Reference regression =====

class TestSvdProjectReference:
    def test_matches_pyena_reference(self, pyena_reference, align_sign):
        result = svd_project(pyena_reference["normalized_matrix"], n_components=2)

        # centered_matrix and singular_values must match exactly
        np.testing.assert_array_equal(
            result.centered, pyena_reference["centered_matrix"]
        )
        np.testing.assert_array_equal(
            result.singular_values, pyena_reference["singular_values"][:2]
        )

        # coords and components must match up to sign
        coords_aligned = align_sign(result.coords, pyena_reference["points_2d"])
        np.testing.assert_allclose(
            coords_aligned, pyena_reference["points_2d"], atol=1e-14
        )

        ref_Vt = pyena_reference["Vt"][:2]
        components = result.components.copy()
        for k in range(2):
            if np.dot(components[k], ref_Vt[k]) < 0:
                components[k] = -components[k]
        np.testing.assert_allclose(components, ref_Vt, atol=1e-14)

    def test_returns_projection_result(self, pyena_reference):
        result = svd_project(pyena_reference["normalized_matrix"], n_components=2)
        assert isinstance(result, ProjectionResult)


class TestMeansRotationReference:
    def test_matches_rena_means_rotation(self, mr_reference, align_sign):
        result = means_rotation(
            mr_reference["centered_input"],
            mr_reference["group_labels"],
            group_a="A_tight", group_b="B_tight",
        )
        # First two columns are MR1, SVD2
        coords_aligned = align_sign(result.points[:, :2], mr_reference["rena_mr_points"])
        diff = np.abs(coords_aligned - mr_reference["rena_mr_points"])
        # rENA-exact verified at machine epsilon
        assert diff.max() < 1e-10, f"max abs diff = {diff.max():.3e}"

    def test_returns_means_rotation_result(self, mr_reference):
        result = means_rotation(
            mr_reference["centered_input"],
            mr_reference["group_labels"],
            group_a="A_tight", group_b="B_tight",
        )
        assert isinstance(result, MeansRotationResult)
        assert result.groups_used == ("A_tight", "B_tight")


# ===== svd_project properties =====

class TestSvdProject:
    def test_singular_values_are_sorted_descending(self):
        rng = np.random.default_rng(0)
        M = rng.random((50, 10))
        result = svd_project(M, n_components=5)
        sv = result.singular_values
        assert all(sv[i] >= sv[i + 1] for i in range(len(sv) - 1))

    def test_components_are_orthonormal(self):
        rng = np.random.default_rng(1)
        M = rng.random((50, 10))
        result = svd_project(M, n_components=5)
        # components are rows of Vt[:k], should be orthonormal
        gram = result.components @ result.components.T
        np.testing.assert_allclose(gram, np.eye(5), atol=1e-12)

    def test_centered_has_zero_mean(self):
        rng = np.random.default_rng(2)
        M = rng.random((50, 10))
        result = svd_project(M, n_components=2)
        np.testing.assert_allclose(result.centered.mean(axis=0), 0, atol=1e-14)

    def test_coords_equal_centered_times_components_T(self):
        rng = np.random.default_rng(3)
        M = rng.random((50, 10))
        result = svd_project(M, n_components=3)
        expected = result.centered @ result.components.T
        np.testing.assert_allclose(result.coords, expected, atol=1e-14)

    def test_n_components_parameter_respected(self):
        rng = np.random.default_rng(4)
        M = rng.random((30, 10))
        for k in (1, 2, 5, 10):
            result = svd_project(M, n_components=k)
            assert result.coords.shape == (30, k)
            assert result.components.shape == (k, 10)


# ===== orthogonal_svd properties =====

class TestOrthogonalSvd:
    def test_first_columns_equal_weights(self):
        """The first k columns of the output must coincide with weights (up to QR's sign)."""
        rng = np.random.default_rng(0)
        data = rng.random((20, 5))
        weights = rng.random((5, 1))
        weights /= np.linalg.norm(weights, axis=0)
        rotation = orthogonal_svd(data, weights)
        # QR may flip sign; compare absolute alignment
        align = abs(np.dot(rotation[:, 0], weights[:, 0]))
        assert align == pytest.approx(1.0, abs=1e-12)

    def test_rotation_is_orthonormal(self):
        rng = np.random.default_rng(1)
        data = rng.random((20, 5))
        weights = rng.random((5, 1))
        weights /= np.linalg.norm(weights, axis=0)
        rotation = orthogonal_svd(data, weights)
        np.testing.assert_allclose(rotation @ rotation.T, np.eye(5), atol=1e-12)


# ===== means_rotation properties =====

class TestMeansRotation:
    def test_mr1_axis_separates_groups(self, mr_reference):
        """The first axis must separate the two groups used for rotation."""
        result = means_rotation(
            mr_reference["centered_input"],
            mr_reference["group_labels"],
            group_a="A_tight", group_b="B_tight",
        )
        labels = mr_reference["group_labels"]
        mr1 = result.points[:, 0]
        # The two group centroids on MR1 must be far apart
        mean_a = mr1[labels == "A_tight"].mean()
        mean_b = mr1[labels == "B_tight"].mean()
        assert abs(mean_a - mean_b) > 0.5  # toy data is well-separated

    def test_rotation_is_orthonormal(self, mr_reference):
        result = means_rotation(
            mr_reference["centered_input"],
            mr_reference["group_labels"],
            group_a="A_tight", group_b="B_tight",
        )
        R = result.rotation
        np.testing.assert_allclose(R @ R.T, np.eye(R.shape[0]), atol=1e-12)

    def test_raises_when_group_missing(self, mr_reference):
        with pytest.raises(ValueError, match="must both be present"):
            means_rotation(
                mr_reference["centered_input"],
                mr_reference["group_labels"],
                group_a="A_tight", group_b="NONEXISTENT",
            )

    def test_raises_when_group_means_identical(self):
        """If both groups have identical centroids, MR1 is undefined."""
        M = np.array([[1.0, 0.0], [1.0, 0.0], [1.0, 0.0], [1.0, 0.0]])
        labels = np.array(["A", "A", "B", "B"])
        with pytest.raises(ValueError, match="undefined"):
            means_rotation(M, labels, group_a="A", group_b="B")
'''

path = PROJECT_ROOT / 'tests' / 'test_projection.py'
path.write_text(projection_test)

import subprocess
result = subprocess.run(
    ['pytest', str(path), '-v'],
    capture_output=True, text=True, cwd=str(PROJECT_ROOT),
)
print(result.stdout[-3500:])
if result.stderr:
    print("=== STDERR ===")
    print(result.stderr[-500:])
print(f"return code: {result.returncode}")
```
