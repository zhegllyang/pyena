```python
import numpy as np
import pandas as pd
import pickle
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams['font.family'] = 'DejaVu Sans'
mpl.rcParams['axes.unicode_minus'] = False

# 1. rpy2 환경 확인 (3.6+ 새 API)
import rpy2.robjects as ro
from rpy2.robjects.packages import importr, isinstalled
from rpy2.robjects import pandas2ri, default_converter
from rpy2.robjects.conversion import localconverter

# 새로운 패턴: localconverter context manager 사용
# (전역 activate 대신, 변환이 필요한 블록에서만 사용)
print(f"R version: {ro.r('R.version.string')[0]}")

# R 기본 동작 테스트
test_result = ro.r('1 + 1')[0]
print(f"R test (1+1): {test_result}")

# 2. 저장한 reference 데이터 로드
df_large = pd.read_csv('./data/toy_data_n90.csv')
with open('./data/pyena_reference_output.pkl', 'rb') as f:
    ref = pickle.load(f)

print(f"\nLoaded toy data: {df_large.shape}")
print(f"  Unique units: {df_large['unit'].nunique()}")
print(f"  Groups: {df_large['group'].unique().tolist()}")
print(f"\nReference keys: {list(ref.keys())}")
print(f"  pyENA points shape: {ref['points_2d'].shape}")
print(f"  Codes: {ref['codes']}")

# 3. rENA 설치 확인
if isinstalled('rENA'):
    rena_ver = ro.r('as.character(packageVersion("rENA"))')[0]
    print(f"\n✓ rENA: {rena_ver}  — ready")
else:
    print(f"\n✗ rENA: not installed")
    print(f"  → 다음 셀에서 설치 시도")
```
